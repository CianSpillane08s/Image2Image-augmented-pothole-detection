```python
import os

# Path to your dataset directory
dataset_dir = '/content/drive/MyDrive/Dataset/Testing/Pictures/Dublin/Dublin Day Light'

# Rename files by removing 'Copy of Copy of Copy of ' prefix
for filename in os.listdir(dataset_dir):
    if filename.startswith("Copy of Copy of Copy of "):  # Match the prefix
        new_filename = filename.replace("Copy of Copy of Copy of ", "")  # Remove prefix
        os.rename(os.path.join(dataset_dir, filename), os.path.join(dataset_dir, new_filename))

print("Renaming completed.")
```
